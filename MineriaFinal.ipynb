```python
import pandas as pd
import numpy as np

np.random.seed(42)
student_ids = [f'{i:06d}' for i in range(172120, 172120 + 20)]
groups = ['Café'] * 10 + ['Agua'] * 10
np.random.shuffle(groups)

coffee_scores = np.random.normal(loc=70, scale=5, size=10).round(0)
coffee_scores = np.clip(coffee_scores, 50, 100).astype(int)

water_scores = np.random.normal(loc=75, scale=5, size=10).round(0)
water_scores = np.clip(water_scores, 50, 100).astype(int)

exam_scores = []
coffee_idx, water_idx = 0, 0
for group in groups:
    if group == 'Café':
        exam_scores.append(coffee_scores[coffee_idx])
        coffee_idx += 1
    else:
        exam_scores.append(water_scores[water_idx])
        water_idx += 1

df_students = pd.DataFrame({
    'Matrícula': student_ids,
    'Grupo': groups,
    'Calificaciones': exam_scores
})

# RESOLUCIÓN PREGUNTA 2: Media de estudiantes que consumen agua
media_agua = df_students[df_students['Grupo'] == 'Agua']['Calificaciones'].mean()
print(f"Pregunta 2 (Media Agua): {media_agua}")

# RESOLUCIÓN PREGUNTA 4: Media de estudiantes que consumen café
media_cafe = df_students[df_students['Grupo'] == 'Café']['Calificaciones'].mean()
print(f"Pregunta 4 (Media Café): {media_cafe}")
```

```python
# Cargar los datos
dfc = pd.read_csv('clientes.csv')
dfo = pd.read_csv('ordenes.csv')

# Realizar el left merge por cliente_id
df_merged = dfc.merge(dfo, on='cliente_id', how='left')

# (a) Número total de filas
total_filas = len(df_merged)

# (b) Monto total por cliente reemplazando NaN por 0
df_merged['monto'] = df_merged['monto'].fillna(0)
monto_por_cliente = df_merged.groupby('cliente_id')['monto'].sum()

# Respuesta específica: Monto total para cliente_id = 1
monto_cliente_1 = monto_por_cliente.loc[1]

print(f"Pregunta 23 (Monto cliente_id=1): {monto_cliente_1}")
```

```python
# Cargar DataFrame solo con columnas necesarias
df_sensores = pd.read_csv('sensores.csv', usecols=['ts', 'temperatura', 'humedad'])

# Convertir 'ts' a datetime manejando formatos mixtos y forzando errores a NaT
df_sensores['ts'] = pd.to_datetime(df_sensores['ts'], format='mixed', errors='coerce')

# (a) ¿Cuántos NaT quedaron en ts?
conteo_nat = df_sensores['ts'].isna().sum()

# Reemplazar valores faltantes
mediana_temp = df_sensores['temperatura'].median()
df_sensores['temperatura'] = df_sensores['temperatura'].fillna(mediana_temp)

media_hum = df_sensores['humedad'].mean()
df_sensores['humedad'] = df_sensores['humedad'].fillna(media_hum)

# (b) Valor final de la mediana de temperatura (redondeado a 1 decimal)
mediana_final = round(mediana_temp, 1)

# (c) Valor final de la media de humedad (redondeado a 3 decimales)
media_final = round(media_hum, 3)

print(f"Pregunta 26 (Respuestas a, b, c): {conteo_nat}, {mediana_final}, {media_final}")
```
