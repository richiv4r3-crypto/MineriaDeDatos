```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/munsell-array-fixed-choice.csv')
df
```

```python
#Porcentage de colores en Inglés
counts = df['english_color'].value_counts()
percentage = df['english_color'].value_counts(normalize=False) * 100
result = pd.DataFrame({'count': counts, 'percentage': percentage.round(2)})
plt.barh(result.index, result['count'])
plt.show()
```

```python
#Porcentage de colores en Español
counts = df['spanish_color'].value_counts()
percentage = df['spanish_color'].value_counts(normalize=False) * 100
resultado = pd.DataFrame({'count': counts, 'percentage': percentage.round(2)})
plt.barh(resultado.index, resultado['count'])
plt.show()
```

```python
#Porcentage de colores en Español
counts = df['tsimane_color'].value_counts()
percentage = df['tsimane_color'].value_counts(normalize=False) * 100
res = pd.DataFrame({'count': counts, 'percentage': percentage.round(2)})
plt.barh(res.index, res['count'])
plt.show()
```

```python
#Correlación Inglés Español
df = pd.read_csv('munsell-array-fixed-choice.csv')

combined = pd.DataFrame({
    'english': df['english_color'].value_counts(),
    'spanish': df['spanish_color'].value_counts()
}).fillna(0)

plt.scatter(combined['english'], combined['spanish'])

for color, row in combined.iterrows():
    plt.annotate(color, (row['english'], row['spanish']))
plt.show()
```

```python
combined = pd.DataFrame({
    'english': df['english_color'].value_counts(),
    'tsimane': df['tsimane_color'].value_counts()
}).fillna(0)

plt.scatter(combined['english'], combined['tsimane'])

for color, row in combined.iterrows():
    plt.annotate(color, (row['english'], row['tsimane']))
plt.show()
```

```python
#Correlación Español Tsimane
combined = pd.DataFrame({
    'spanish': df['spanish_color'].value_counts(),
    'tsimane': df['tsimane_color'].value_counts()
}).fillna(0)

plt.scatter(combined['spanish'], combined['tsimane'])

for color, row in combined.iterrows():
    plt.annotate(color, (row['spanish'], row['tsimane']))
plt.show()
```
