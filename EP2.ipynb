#Pregunta 2

```python
df = pd.read_csv("items.csv")
```

```python
#Con eval(), crea la columna importe = precio * cantidad.
df['importe'] = df.eval('precio * cantidad')
print(df.head(3))
```

```python
#Con query(), filtra solo categoria == 'A' y guarda en dfa.
dfa = df.query('categoria == "A"')
print(dfa.head(3))
```

```python
#Con groupby(), calcula el promedio de importe por categoria en el DataFrame original df (no filtrado).
df.groupby('categoria')['importe'].mean()
```

```python
#¿Cuál es el promedio de importe de la categoría A (en el DataFrame original, no el filtrado)? Redondea a 3 decimales.
round(df.query('categoria == "A"')['importe'].mean(),3)
```

#Pregunta 3

```python
dfc = pd.read_csv("clientes.csv")
dfo = pd.read_csv("ordenes.csv")
```

```python
#Realiza un left merge de dfc con dfo por cliente_id, manteniendo todas las filas de dfc.
dfc = pd.merge(dfc, dfo, on='cliente_id', how='left')
print(dfc)
```

```python
#Calcula: (a) Número total de filas del resultado del left merge.
dfc.shape[1]
```

```python
#Calcula: (b) Monto total por cliente (suma de monto) y reemplaza NaN por 0.
dfc['monto'] = dfc['monto'].fillna(0)
dfc.groupby('cliente_id')['monto'].sum()
```

```python
#¿Cuál es el monto total por cliente cliente_id=1?
dfc.query('cliente_id == 1')['monto'].sum()
```

#Pregunta 4

```python
#Carga el CSV en df, selecciona solo columnas: ts, temperatura, humedad.
df = pd.read_csv("datos.csv", usecols=['ts','temperatura','humedad'])
```

```python
#Convierte ts a datetime manejando formatos mixtos (usa format='mixed' y errors='coerce').
df['ts'] = pd.to_datetime(df['ts'], format='mixed', errors='coerce')
```

```python
#Reemplaza valores faltantes en temperatura con la mediana y en humedad con la media (usa fillna).
df['temperatura'] = df['temperatura'].fillna(df['temperatura'].median())
```

```python
#¿Cuántos NaT quedaron en ts?
total_nan = df['ts'].isna().sum()
print(total_nan)
```

```python
#Valor final de la mediana de temperatura tras imputación (redondea a 1 decimal).
final_median_temp = round(df['temperatura'].median(), 1)
print(final_median_temp)
```

```python
#Valor final de la media de humedad tras imputación (redondea a 3 decimales).
final_mean_hum = round(df['humedad'].mean(), 3)
print(final_mean_hum)
```

```python
#Formato de entrega: tres números (en orden a, b, c):
print(f"{total_nan}, {final_median_temp}, {final_mean_hum}")
```
